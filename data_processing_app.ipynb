```python
df = pd.read_csv("students.csv")

print("Original Dataset:")
display(df)
```

```python
def clean_data(df):
    # Remove duplicate rows
    df = df.drop_duplicates()

    # Convert score values to numbers
    df["score"] = pd.to_numeric(df["score"], errors="coerce")

    # Treat scores outside 0–100 as invalid
    df.loc[~df["score"].between(0, 100), "score"] = np.nan

    # Fill missing scores with the median of valid scores
    median_score = df["score"].median()
    df["score"] = df["score"].fillna(median_score)

    return df


cleaned_df = clean_data(df.copy())

print("Cleaned Dataset:")
display(cleaned_df)
```

```python
# Get the scores as a NumPy array
scores = cleaned_df["score"].to_numpy()

# Calculate statistics using NumPy
average = np.mean(scores)
highest = np.max(scores)
lowest = np.min(scores)
standard_deviation = np.std(scores)

print("Student Score Analysis")
print("----------------------")
print("Average Score:", round(average, 2))
print("Highest Score:", highest)
print("Lowest Score:", lowest)
print("Standard Deviation:", round(standard_deviation, 2))
```

```python
cleaned_df.to_csv("cleaned_students.csv", index=False)

print("Cleaned dataset saved successfully!")
```
